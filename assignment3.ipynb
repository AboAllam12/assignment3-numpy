```python
#1
import numpy as np

arr=np.random.randint(1,100,(3,3))
print(f"Array:\n{arr}")
min_arr=np.min(arr)
max_arr=np.max(arr)
norm_arr=(arr-min_arr)/(max_arr-min_arr)
print(f"Normalized Array:\n{norm_arr}")
```

```python
#2
import numpy as np

arr=np.random.randint(0,20,10)
print(f"Array:\n{arr}")
x=int(input("Enter x: "))
Closest=arr[np.abs(arr-x).argmin()]
print(f"Closest value to {x} in the array is: {Closest}")
```

```python
#3
import numpy as np

arr=np.random.randint(0,100,(5,5))
print(f"Array:\n{arr}")
arr_sorted=np.sort(arr,axis=0)
print(f"Sorted Array:\n{arr_sorted}")
second_largest=arr_sorted[-2,:]
print(f"Second largest value in each column:\n{second_largest}")
```

```python
#4
import numpy as np

arr=np.random.randint(0,100,(6,4))
print(f"Array:\n{arr}")
arr_sum=np.sum(arr,axis=1)
print(f"Sum of each row:\n{arr_sum}")
even=np.where(arr_sum%2==0)
print(f"Indices of even sums:\n{even[0]}")
```

```python
#5
import numpy as np

arr=np.random.randint(1,51,30)
print(f"Array:\n{arr}")
arr[(arr%3==0)|(arr%5==0)] =-1
print(f"Modified Array:\n{arr}")
```

```python
#6
import numpy as np

arr=np.random.randint(1,21,(3,3))
print(f"Array:\n{arr}")
arr_mean=np.mean(arr,axis=1,keepdims=True)
print(f"Mean of each row:\n{arr_mean}")
result=arr-arr_mean
print(f" Array after subtracting mean from each row:\n{result}")
```
